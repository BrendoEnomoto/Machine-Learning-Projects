#Introdução
O código usa o algoritmo Random Forest para classificar se uma pessoa tem uma renda anual acima ou abaixo de 50 mil dólares, com base nos dados do censo. O algoritmo Random Forest é um algoritmo de aprendizado de máquina de conjunto que cria um conjunto de árvores de decisão e combina suas previsões para tomar uma decisão final.


## Importação da base de dados

```python
import pandas as pd

census = pd.read_csv('census.csv')
```

```python
X = census.iloc[:, 0:14].values
y = census['income'].values
```

## Pré-processamento de dados:
Os dados do censo são carregados e pré-processados
​​usando LabelEncoder e StandardScaler. LabelEncoder converte variáveis ​​categóricas em numéricas e StandardScaler dimensiona os recursos numéricos para terem média zero e variância unitária.

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

X[:,1] = encoder.fit_transform(X[:,1])
X[:,3] = encoder.fit_transform(X[:,3])
X[:,5] = encoder.fit_transform(X[:,5])
X[:,6] = encoder.fit_transform(X[:,6])
X[:,7] = encoder.fit_transform(X[:,7])
X[:,8] = encoder.fit_transform(X[:,8])
X[:,9] = encoder.fit_transform(X[:,9])
X[:,13] = encoder.fit_transform(X[:,13])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X = scaler.fit_transform(X)
```

## Divisão de dados:
 Os dados são divididos em conjuntos de treinamento e teste. O conjunto de treinamento é usado para treinar o modelo Random Forest, e o conjunto de teste é usado para avaliar seu desempenho.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

## Treinamento do modelo:
O modelo Random Forest é treinado usando os dados de treinamento. Isso envolve a criação de um conjunto de árvores de decisão, onde cada árvore é treinada em um subconjunto aleatório dos dados de treinamento.

```python
from sklearn.ensemble import RandomForestClassifier

random_forest_census = RandomForestClassifier(n_estimators=100, criterion='entropy', random_state = 0)

random_forest_census.fit(X_train, y_train)
```

## Previsão:
 O modelo treinado é usado para fazer previsões sobre os dados de teste. Isso é feito combinando as previsões de todas as árvores de decisão no conjunto.

```python
pred = random_forest_census.predict(X_test)
```

## Avaliação:
O desempenho do modelo é avaliado usando métricas como precisão, precisão e revocação. Essas métricas nos dizem o quão bem o modelo é capaz de prever a renda de uma pessoa.

```python
from sklearn.metrics import accuracy_score, classification_report

print(classification_report(y_test, pred))
```
