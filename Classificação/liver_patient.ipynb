```python
map_v2 = {
    'Female': 0,
    'Male': 1
}
df['V2'] = df['V2'].map(map_v2)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import uniform
from scipy.stats import randint
from sklearn.metrics import f1_score, make_scorer
```

SVC e o Random Forest Classifier, com Random Search para
sintonia de hiperparâmetros e validação cruzada estratificada.

# Implementação SVC

```python
svc_classifier = SVC()

cv_strategy = StratifiedKFold(n_splits=10)

f1 = make_scorer(f1_score)

svc_parameters = {
    'kernel': ['sigmoid', 'poly', 'rbf', ],
    'C': uniform(1, 10)
}

svc_random_search = RandomizedSearchCV(
    estimator=svc_classifier,
    param_distributions=svc_parameters,
    n_iter=5,
    cv=cv_strategy,
    scoring=f1,
    random_state=5762
)
svc_random_search.fit(df.drop('Class', axis=1), df['Class'])
```

```python
svc_random_search.best_score_
```

```python
svc_random_search.best_params_
```

```python
svc_random_search.best_estimator_
```

##Implementação Random Forest Classifier

```python
rf_classifier = RandomForestClassifier(random_state = 5762)

cv_strategy = StratifiedKFold(n_splits=10)

f1 = make_scorer(f1_score)

rf_parameters = {
    'n_estimators': randint(10, 1000),
    'bootstrap' : [True, False],
    'criterion': ['gini', 'entropy']
}

rf_random_search = RandomizedSearchCV(
    estimator=rf_classifier,
    param_distributions=rf_parameters,
    n_iter=5,
    cv=cv_strategy,
    scoring=f1,
    random_state=5762
)
rf_random_search.fit(df.drop('Class', axis=1), df['Class'])
```

```python
rf_random_search.best_score_
```

```python
rf_random_search.best_params_
```

```python
rf_random_search.best_estimator_
```
