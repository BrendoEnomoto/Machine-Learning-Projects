**Aplique o modelo de normalização MinMaxScaler com os valores "default" para os dados de entrada. Qual é o menor valor para a variável "fixed acidity"?**

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler( )

red_wine_df_scaled = scaler.fit_transform(red_wine_df)

red_wine_df_scaled = pd.DataFrame(red_wine_df_scaled, columns=red_wine_df.columns)

red_wine_df_scaled['fixed acidity'].max()
```

```python
from sklearn.model_selection import train_test_split

X = red_wine_df.drop('quality', axis=1)
y = red_wine_df['quality']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=1)
```

**Implementação do KNN**

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

clf_KNN = KNeighborsClassifier(n_neighbors=5)

clf_KNN.fit(X_train, y_train)

y_pred = clf_KNN.predict(X_test)

accuracy_score(y_test, y_pred)
```

**Implementação Algoritmo Árvore de Decisão**

```python
from sklearn.tree import DecisionTreeClassifier

clf_arvore = DecisionTreeClassifier()

clf_arvore.fit(X_train, y_train)

y_pred = clf_arvore.predict(X_test)

accuracy_score(y_test, y_pred)
```

**Implementação Algoritmo Floresta Randônica**

```python
from sklearn.ensemble import RandomForestClassifier

clf_floresta = RandomForestClassifier(max_depth=10, random_state=1)

clf_floresta.fit(X_train, y_train)

y_pred = clf_floresta.predict(X_test)

accuracy_score(y_test, y_pred)
```

**Implementação Algoritmo SVM**

```python
from sklearn.svm import SVC

clf_svm=SVC(gamma='auto',kernel='rbf')

clf_svm.fit(X_train, y_train)

y_pred = clf_svm.predict(X_test)

accuracy_score(y_test, y_pred)
```

```python
from sklearn.neural_network import MLPClassifier

clf_mlp = MLPClassifier(alpha=1e-5, hidden_layer_sizes=(5,5), random_state=1)

clf_mlp.fit(X_train, y_train)

y_pred = clf_mlp.predict(X_test)

accuracy_score(y_test, y_pred)
```

**Modifique o dataset para um sistema binário que considere vinhos com nota maior do que 5 (6, 7, 8) sejam bons e menores ou igual a 5 ruins (5, 4, 3). Aplique o modelo da floresta randômica. Qual é o valor mais próximo da acurácia do modelo para esse novo dataset?**

```python
red_wine_df = pd.read_csv('winequality-red.csv', sep=';')
red_wine_df['quality'] = red_wine_df['quality'].apply(lambda x: 1 if x > 5 else 0)
red_wine_df['quality'].value_counts()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler( )

red_wine_df_scaled = scaler.fit_transform(red_wine_df)

red_wine_df_scaled = pd.DataFrame(red_wine_df_scaled, columns=red_wine_df.columns)
```

```python
from sklearn.model_selection import train_test_split

X = red_wine_df.drop('quality', axis=1)
y = red_wine_df['quality']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=1)
```

```python
from sklearn.ensemble import RandomForestClassifier

clf_floresta = RandomForestClassifier(max_depth=10, random_state=1)

clf_floresta.fit(X_train, y_train)

y_pred = clf_floresta.predict(X_test)

accuracy_score(y_test, y_pred)
```
